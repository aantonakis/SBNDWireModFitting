```python
values, edges = h_mc.to_numpy()
norm_values = values / np.sum(values)
plt.stairs(norm_values, edges, label="MC", color="blue", linewidth=2)

values, edges = h_data.to_numpy()   
norm_values = values / np.sum(values)
plt.stairs(norm_values, edges, label="Data", color="black", linewidth=2)

plt.plot([itm_mc, itm_mc], [0, 1.1*max(norm_values)], color="blue", linestyle="--", label="MC ITM")
plt.plot([itm_data, itm_data], [0, 1.1*max(norm_values)], color="black", linestyle="--", label="Data ITM")
plt.legend(fontsize=14)

if h_mc.member("fXaxis").member("fTitle") == "Q":
    xname = "Hit Q [ADC]"

elif h_mc.member("fXaxis").member("fTitle") == "W":
    xname = "Hit Width [ticks]"

plt.xlabel(xname, fontsize=16)
plt.ylabel("Area Normalized", fontsize=16)
plt.xlim([0, 1000])
plt.show()
```
